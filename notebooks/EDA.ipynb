```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('churn_cleaned_data.csv')
```

```python
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. فحص عدم التوازن في الـ Target (Churn)
plt.figure(figsize=(6, 4))
sns.countplot(x='Churn', data=df)
plt.title('Target Distribution (Churn vs Non-Churn)')
plt.show()

print(df['Churn'].value_counts(normalize=True) * 100)

# 2. فحص مصفوفة الارتباط (Correlation Matrix) مع الـ Churn
plt.figure(figsize=(10, 8))
df.corr()['Churn'].sort_values().plot(kind='barh')
plt.title('Feature Correlation with Churn')
plt.xlabel('Correlation Coefficient')
plt.show()
```

```python
# رسم التوزيعات للقيم الرقمية الرئيسية
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(df['TenureMonths'], kde=True, ax=axes[0], color='teal')
axes[0].set_title('Tenure Months Distribution')

sns.histplot(df['MonthlyCharges'], kde=True, ax=axes[1], color='coral')
axes[1].set_title('Monthly Charges Distribution')

sns.histplot(df['TotalCharges'], kde=True, ax=axes[2], color='purple')
axes[2].set_title('Total Charges Distribution')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Tenure vs Churn
sns.boxplot(x='Churn', y='TenureMonths', data=df, ax=axes[0], palette='Set2')
axes[0].set_title('Tenure Months vs Churn')

# Monthly Charges vs Churn
sns.boxplot(x='Churn', y='MonthlyCharges', data=df, ax=axes[1], palette='Set2')
axes[1].set_title('Monthly Charges vs Churn')

plt.tight_layout()
plt.show()
```
